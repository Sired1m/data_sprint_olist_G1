```python
#Why are some customers rating Olist 1 star (high dissatisfaction)?
#How can we convert "highly unsatisfied" customers into satisfied ones?
#How can we improve 4-star reviews to 5 stars?



one = df[df['review_score'] == 1].copy()

def status(r):
    if pd.isna(r['order_delivered_customer_date']):
        return 'Not delivered'   # no delivery date
    if r['late']:
        return 'Late'
    return 'On time / early'

one['delivery_status'] = one.apply(status, axis=1)

print(one['delivery_status'].value_counts())
one['delivery_status'].value_counts().plot(kind='pie', autopct='%.0f%%', title='1-star orders: delivery status')


# 1-star reviews that have a comment
print('1-star with comment:', one['review_comment_message'].notna().sum(), 'of', len(one))
```

```python
# 1-star only, join title and message
one = reviews_n[reviews_n['review_score'] == 1].copy()
one['text'] = (one['eng_review_comment_title'].fillna('') + ' ' + one['eng_review_comment_message'].fillna('')).str.lower()

# categories and keywords (first match wins, so the order matters)
categories_idk = {
    'Not received': ['not received', 'never received', 'did not arrive', "didn't arrive", 'not arrived', 'never arrived',
                     'not delivered', 'did not receive', "didn't receive", 'have not received', "haven't received",
                     "hasn't arrived", 'has not arrived', 'not receive', 'receive my', 'never came', "didn't come",
                     'not come', 'not arrive', "wasn't delivered", 'was not delivered', "wasn't actually delivered",
                     'neither has been delivered', 'never got', 'not yet', 'no delivery', 'lost', 'still have not'],
    'Late delivery': ['late', 'delay', 'deadline', 'still waiting', 'waiting', 'wait', 'days', 'weeks', 'month',
                      'so long', 'long time', 'too long', 'to arrive', 'absurd amount of time', 'slow', 'took',
                      'forever', 'promised', 'estimated', "didn't call"],
    'Damaged/defect': ['broken', 'broke', 'damaged', 'defect', 'faulty', 'does not work', "doesn't work", 'not working',
                       'stopped working', 'torn', 'cracked', 'flaws', 'dent', 'scratch', 'dirty', 'opened', 'leak',
                       'caught fire', 'burn', 'malfunction', 'does not turn', 'not turn on', 'dead', 'used'],
    'Wrong item': ['wrong', 'different', 'another product', 'another', 'not what', 'not the one', 'not the same',
                   'incomplete', 'missing', 'only one', 'only 1', 'but only', 'instead', 'they sent', 'sent me',
                   'came only', 'as described', 'does not match', 'color', 'colour', 'size', 'lack', 'kit',
                   "wasn't the one i chose"],
    'Seller/Service': ['seller', 'service', 'support', 'contact', 'refund', 'cancel', 'return', 'answer', 'response',
                       'respond', 'reply', 'store', 'shop', 'customer', 'company', 'email', 'call', 'phone', 'money',
                       'charged', 'paid', 'payment', 'invoice', 'exchange', 'complaint', 'protocol', 'information'],
    'Bad quality': ['quality', 'bad', 'poor', 'terrible', 'horrible', 'awful', 'worst', 'fake', 'counterfeit',
                    'disappointed', 'disappointing', 'disappointment', 'useless', 'waste', 'cheap', 'weak', 'fragile',
                    'not recommend', "don't recommend", 'do not recommend', 'do not buy', 'never again', 'regret',
                    'lousy', 'not good', 'nothing'],
}

def categorize(t):
    t = t.strip(' .!')
    if t in ['', 'no comment', 'no comments']:
        return 'No comment'
    for name, words in categories_idk.items():
        if any(w in t for w in words):
            return name
    return 'Other'

one['category'] = one['text'].apply(categorize)

# see the result (count and %)
print(one['category'].value_counts())


one['category'].value_counts().plot(kind='barh', title='1-star reviews by reason')
```

```python
#How can we improve 4-star reviews to 5 stars?

# group orders by how early or late they arrived
groups = pd.cut(deliv['days_late'], [-20, -15, -8, 0, 20], labels=['15+ days early', '8-14 days early', '0-7 days early', 'Late'])

# % of 5-star and 4-star in each group
five = (deliv['review_score'] == 5).groupby(groups).mean() * 100
four = (deliv['review_score'] == 4).groupby(groups).mean() * 100

pd.DataFrame({'5-star %': five, '4-star %': four}).plot(kind='bar', title='Delivery time vs 4 and 5 stars')
```

```python
#Does late delivery cause low reviews, or are there other factors involved? (Correlation ≠ Causation)

groups = pd.cut(deliv['days_late'], [-20, 0, 3, 7, 14, 20], labels=['On time', '1-3 late', '4-7 late', '8-14 late', '15+ late'])
deliv.groupby(groups)['review_score'].mean().plot(kind='bar', title='Average score by days late')


# correlation: shows a link, NOT proof of cause
print('Correlation:', round(deliv['days_late'].corr(deliv['review_score']), 2))

# other factors: 1-star orders that were NOT late
one = deliv[deliv['review_score'] == 1]
print('1-star but on time:', round((~one['late']).mean() * 100), '%')
```
